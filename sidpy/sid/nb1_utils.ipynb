```python
import numpy as np
from sidpy.sid.dataset import Dataset
from sidpy.sid.dimension import Dimension
```

```python
data = np.random.rand(4, 5, 6, 7)  # e.g., X=4, Y=5, Voltage=6, Time=7 
```

```python
data.shape
```

```python
data[0, 0, 0, :]
```

```python
def create_sidpy_dataset_dynamic(data):
    dset = Dataset.from_array(data, title="Dynamic Dataset")

    for i in range(data.ndim):
        print(f"\n--- Axis {i} ---")
        name = input("Axis name: ")
        quantity = input("Quantity: ")
        units = input("Units: ")
        values = np.arange(data.shape[i])  # default
        
        dim = Dimension(values, name=name, quantity=quantity, units=units)
        dset.set_dimension(i, dim)
    
    return dset
```

```python
dset = create_sidpy_dataset_dynamic(data)
```
